Le travail : **5 000 documents**, 600 tokens en entrée et 120 en sortie par document. Exactement le même travail pour les trois modèles.

Remplis le dictionnaire `couts` (une clé par taille de modèle, la valeur en euros), puis :

- `facteur` — combien de fois le grand coûte le petit

Résultats attendus : ≈ `5.46` €, `11.07` €, `27.60` € — et un facteur ≈ `5.05`.

<details><summary>Indice</summary>

Une boucle `for modele in tarifs:` et la même formule qu'à l'exercice 5. Pour le moins cher : `min(couts, key=couts.get)`.

<details><summary>Solution</summary>

```python
NB_DOCS, ENTREE, SORTIE = 5_000, 600, 120

for modele, t in tarifs.items():
    couts[modele] = (NB_DOCS * ENTREE / 1e6) * t["entree"] + (NB_DOCS * SORTIE / 1e6) * t["sortie"]

le_moins_cher = min(couts, key=couts.get)      # "petit"
facteur = couts["grand"] / couts["petit"]      # ≈ 5.05
```

Cinq fois le prix pour le même travail. La question n'est jamais « quel est le meilleur modèle ? » mais « **le petit suffit-il pour cette tâche ?** ». Pour classer, extraire, reformater : très souvent oui.

```python
def plus_gros_modele(budget, nb_docs, tokens_entree, tokens_sortie):
    for modele in ("grand", "moyen", "petit"):        # du plus gros au plus petit
        if cout_euros(nb_docs, tokens_entree, tokens_sortie, modele) <= budget:
            return modele
    return None                                        # même le petit ne rentre pas
```

Deux détails qui font la différence : l'ordre de la boucle (le premier qui rentre est le plus gros possible) et le `<=` plutôt que `<` (un budget de 20,32 € paie exactement 20,32 € de travail).

Le cas `None` n'est pas un échec du code, c'est une **information** : il faut réduire le nombre de documents, raccourcir les réponses, ou demander plus de budget. Mieux vaut l'apprendre maintenant qu'à la moitié du traitement.

## Exercice 11 ⭐⭐⭐ · Quel langage pour quelle équipe

Le choix d'un langage n'est jamais seulement technique : il dépend de ce qu'on livre et de qui va le faire tourner.

Écris `langage_pour(besoin, va_en_production, equipe_connait_r)` qui applique ces règles, **dans cet ordre** :

1. `besoin == "requeter une base"` → `"SQL"`
2. `besoin == "page web"` → `"JavaScript"`
3. `besoin == "moteur de calcul"` → `"Rust"`
4. `besoin == "statistiques"`, **et** ça ne va pas en production, **et** l'équipe connaît R → `"R"`
5. dans tous les autres cas → `"Python"`

La règle 4 est celle qui compte : de la statistique qui part en production, ou une équipe qui ne connaît pas R, c'est Python — parce que le prototype R devrait être réécrit, et qu'on ferait deux fois le travail.

<details><summary>Indice</summary>

Une suite de `if ... : return ...`. Pour la règle 4, une seule condition avec deux `and`. Le `return "Python"` final attrape tout le reste.

</details>

````python
# À toi
def langage_pour(besoin, va_en_production, equipe_connait_r):
    return None

cas = [
    ("statistiques",     False, True),    # une étude à publier, équipe de statisticiens
    ("statistiques",     True,  True),    # la même chose, mais ça part en production
    ("statistiques",     False, False),   # une étude, mais personne ne connaît R
    ("requeter une base", True, False),
    ("page web",         True,  False),
    ("moteur de calcul", True,  False),
    ("classer des avis", True,  True),
]
for besoin, prod, connait_r in cas:
    print(f"{langage_pour(besoin, prod, connait_r)!s:12} ← {besoin} (prod={prod}, R={connait_r})")
````

````python
verifier("Exercice 11 · étude de stat, équipe R", lambda: langage_pour("statistiques", False, True) == "R")
verifier("Exercice 11 · stat qui part en production", lambda: langage_pour("statistiques", True, True) == "Python")
verifier("Exercice 11 · stat sans équipe R", lambda: langage_pour("statistiques", False, False) == "Python")
verifier("Exercice 11 · base de données", lambda: langage_pour("requeter une base", True, False) == "SQL")
verifier("Exercice 11 · page web", lambda: langage_pour("page web", True, False) == "JavaScript")
verifier("Exercice 11 · moteur de calcul", lambda: langage_pour("moteur de calcul", True, False) == "Rust")
verifier("Exercice 11 · le cas par défaut", lambda: langage_pour("classer des avis", True, True) == "Python")
````

<details><summary>Solution</summary>

```python
def langage_pour(besoin, va_en_production, equipe_connait_r):
    if besoin == "requeter une base":
        return "SQL"
    if besoin == "page web":
        return "JavaScript"
    if besoin == "moteur de calcul":
        return "Rust"
    if besoin == "statistiques" and not va_en_production and equipe_connait_r:
        return "R"
    return "Python"
```

Les trois premières règles sont des contraintes : les données sont dans une base, ça doit vivre dans un navigateur, ça doit tenir en 3 millisecondes. On ne discute pas.

La quatrième est un **arbitrage**, et c'est le seul endroit où R apparaît : livrable = analyse, pas de production, l'équipe est à l'aise. Change un seul de ces trois éléments et Python reprend la main — non parce qu'il est meilleur en statistique, mais parce que le prototype R finirait réécrit.

</details>

Douze exercices, quatre réflexes :

1. **Le langage se choisit sur une contrainte**, pas sur une habitude : les données sont dans une base, ça vit dans un navigateur, ça doit tenir en 3 millisecondes, c'est une analyse à publier. Sinon, Python.
