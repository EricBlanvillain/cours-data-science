## 1. Pourquoi Python

Un langage de programmation, c'est une langue : il n'y en a pas de « meilleure », il y en a une qui est parlée là où tu veux aller.

En data et en IA, six langages reviennent tout le temps. Voici honnêtement ce que chacun sait faire, là où il coince, et **qui s'occupe de la mémoire** quand ton programme tourne — une colonne qu'on ne montre presque jamais, alors qu'elle explique la moitié des différences.

| Langage | Ce qu'il sait faire | Sa limite | Gestion de la mémoire | Qui l'utilise |
|---|---|---|---|---|
| **Python** | Tout le chemin : lire un fichier, nettoyer, tracer, entraîner un modèle, servir une API, écrire un script | Lent quand on écrit des boucles à la main | **Ramasse-miettes** : tu n'y penses jamais | Data scientists, ingénieurs IA, chercheurs. C'est le langage de pandas, scikit-learn, PyTorch |
| **R** | Statistiques et graphiques, avec une élégance que Python n'a pas (`ggplot2`) | Sort mal du monde de la stat : industrialiser un modèle R est pénible | **Ramasse-miettes**, avec une copie du tableau à chaque modification | Statisticiens, biostatistique, recherche académique, marketing |
| **SQL** | Interroger des données déjà rangées dans une base, même des milliards de lignes | Ne sait *que* ça : pas de graphique, pas d'entraînement de modèle | **Ce n'est pas ton problème** : le moteur de la base s'en occupe | Absolument tout le monde en entreprise. C'est le langage des bases de données |
| **C / C++** | Aller vite. Très vite. Contrôle total de la machine | Un oubli de mémoire = fuite ou plantage. Peu de bibliothèques data modernes | **À la main** : `malloc` pour prendre, `free` pour rendre | Ceux qui ont écrit les moteurs : NumPy, pandas, la plupart des bibliothèques scientifiques |
| **Rust** | Aussi vite que le C, sans ses accidents de mémoire | Sévère à l'apprentissage : le compilateur refuse beaucoup | **Garantie à la compilation** : ni ramasse-miettes, ni `free` à écrire | Ceux qui écrivent les moteurs d'aujourd'hui : Polars, les serveurs de modèles |
| **JavaScript** | Faire vivre une page web : boutons, graphiques interactifs, tableaux de bord | Écosystème data et IA très pauvre à côté de Python | **Ramasse-miettes** (celui du navigateur) | Développeurs web, tout ce qui est « visible dans un navigateur » |

Le point à retenir : **on choisit un langage pour son écosystème, pas pour sa vitesse.**

Un langage tout seul ne fait rien. Ce qui compte, c'est la pile de bibliothèques déjà écrites, testées et documentées par des milliers de gens, que tu peux appeler en une ligne. Python a environ **600 000 paquets** sur [PyPI](https://pypi.org). Charger un CSV, tracer une courbe, entraîner une forêt aléatoire : trois lignes, parce que quelqu'un a écrit les 30 000 lignes d'en dessous à ta place.

Rust est 50 fois plus rapide que Python. Mais si tu dois écrire toi-même le chargement de CSV, l'algorithme de régression et la bibliothèque de graphiques, tu auras mis trois mois là où Python te demande un après-midi. **Ton temps coûte plus cher que le temps de calcul** — c'est la vraie raison.

### R, sérieusement

R est le langage qu'on cite le plus vite et qu'on traite le plus mal. Il mérite mieux qu'une ligne de tableau, parce que sur son terrain **il gagne**, et que ce terrain est immense.

**Là où R est le meilleur outil disponible :**

- **Les statistiques.** Les modèles linéaires et mixtes, les tests, l'analyse de survie, les séries temporelles : R les propose en une ligne, avec une sortie déjà lisible par un relecteur. En Python, il faut souvent chercher, assembler, et parfois se contenter de moins.
- **[`ggplot2`](https://ggplot2.tidyverse.org/).** Ce n'est pas « une bibliothèque de graphiques de plus » : c'est une **grammaire** — des données, une correspondance entre variables et propriétés visuelles, des couches. Une fois comprise, on décrit un graphique au lieu de le dessiner. `matplotlib` n'a pas d'équivalent.
- **La recherche académique et la biostatistique.** Un essai clinique, une étude d'épidémiologie, un article de génomique : les méthodes de référence sont publiées **en R**, sur [CRAN](https://cran.r-project.org/) et Bioconductor, souvent des années avant d'exister ailleurs. Dans ces milieux, R n'est pas un choix, c'est la langue commune.

**Là où R coince, et il faut le dire aussi :**

- **L'industrialisation.** Mettre un modèle R en production — le faire répondre à des milliers d'appels par jour, dans le système de l'entreprise, surveillé et redémarré tout seul — est nettement plus pénible qu'en Python. Les outils existent, mais les équipes d'ingénierie qui doivent faire tourner la chose ne parlent pas R. Le scénario classique : le modèle est **prototypé en R** par le statisticien, puis **réécrit en Python** pour la production. Deux fois le travail, et deux occasions de se tromper.
- **Tout ce qui n'est pas de la statistique.** Réseaux de neurones, traitement d'images, appel d'un modèle de langage, scraping web, mise en place d'une API : à chaque fois, l'écosystème R existe mais reste en retrait, moins maintenu, moins documenté. On y perd le confort exact qu'on était venu chercher.

**La règle honnête :** si ton livrable est une *analyse* — un article, un rapport, une étude — R est un excellent choix, parfois le meilleur. Si ton livrable est un *système* — quelque chose qui tourne tout seul et que quelqu'un d'autre appelle — prends Python.

## 2. Qui gère la mémoire, et ce que ça te coûte

Quand ton programme tourne, il réclame de la mémoire à la machine : un tableau, une chaîne de caractères, un modèle. À un moment, il faut la **rendre**. La question « qui s'en charge ? » a exactement trois réponses, et chacune se paie dans une monnaie différente.

**1 · Le ramasse-miettes — Python, R, JavaScript.** Le langage compte combien de variables pointent sur chaque objet. Quand plus personne ne pointe dessus, il est libéré, tout seul, sans que tu écrives quoi que ce soit.
*Ce que ça te coûte :* du temps de calcul — ce comptage tourne en permanence — et le fait que tu ne décides pas *quand* la mémoire est rendue. Et ça ne te protège pas de tout : un objet oublié dans une liste qui grossit n'est jamais libéré, parce que quelqu'un pointe encore dessus. C'est le prix de la tranquillité.

**2 · À la main — C, C++.** Tu demandes la mémoire (`malloc`), tu la rends (`free`). Rien ne tourne en arrière-plan : c'est aussi rapide que possible, et tu contrôles tout, à l'octet près.
*Ce que ça te coûte :* ton attention, en permanence. Un `free` oublié, c'est une **fuite** : le programme grossit jusqu'à saturer la machine. Un `free` fait deux fois, ou une lecture après libération, c'est un plantage — ou une faille de sécurité. Une grande partie des vulnérabilités graves des trente dernières années vient exactement de là.

**3 · Garantie à la compilation — Rust.** Le [compilateur suit qui possède quoi](https://doc.rust-lang.org/book/ch04-00-understanding-ownership.html) et insère lui-même la libération au bon endroit. Aucun ramasse-miettes à l'exécution, aucun `free` à écrire, et les bugs de mémoire deviennent **impossibles** : ils ne compilent pas.
*Ce que ça te coûte :* la sévérité. Le compilateur **refuse ton programme** tant qu'il a le moindre doute, même quand tu es sûr d'avoir raison. Les premières semaines, on se bat avec lui. Ensuite, on ne se bat plus jamais avec un plantage en production.

En une phrase : **Python et R achètent ta tranquillité avec de la performance, le C achète la performance avec ton attention, Rust achète les deux avec ta patience à la compilation.** Il n'y a pas de gagnant : il y a une facture, et tu choisis dans quelle monnaie tu la règles.

Regardons le ramasse-miettes de Python en train de travailler.

````python
import sys

class Bloc:
    """Un objet qui annonce le moment exact où Python le libère."""
    def __init__(self, nom):
        self.nom = nom
    def __del__(self):
        print(f"   → libéré automatiquement : {self.nom}")

bloc = Bloc("mes données")
print("après `bloc = Bloc(...)`  : ", sys.getrefcount(bloc) - 1, "référence(s)")

copie = bloc
print("après `copie = bloc`      : ", sys.getrefcount(bloc) - 1, "référence(s)")

del copie
print("après `del copie`         : ", sys.getrefcount(bloc) - 1, "référence(s)")

print("on supprime la dernière référence :")
del bloc
print("Personne ne pointait plus dessus : Python a rendu la mémoire, sans qu'on demande.")
````

**Une réserve, tout de suite.** Si la libération se voit ici immédiatement, c'est parce que CPython — l'implémentation de Python que tu utilises — compte les références et libère dès que le compteur tombe à zéro. Mais **le moment exact n'est garanti par le langage** : une autre implémentation, ou le ramasse-miettes sur des objets qui se pointent l'un l'autre, peut libérer plus tard. On n'écrit donc jamais un programme dont le fonctionnement dépend de l'instant où `__del__` se déclenche (pour fermer un fichier ou une connexion, on utilise `with`). C'est une démonstration, pas une technique à réutiliser.

En C, cette libération, c'est **toi** qui l'écris. En Rust, c'est le compilateur qui l'a écrite pour toi, avant même que le programme démarre.

## 3. Le coût d'un modèle

Dernière question, et la plus concrète : **combien ça coûte ?**

Quand tu appelles un modèle par API, tu paies **au token** — ces morceaux de mots qu'on découpera à la séance 9. Ordre de grandeur utile : **1 token ≈ 4 caractères**, soit environ **750 mots pour 1 000 tokens** en français.

Tu paies deux fois, à des tarifs différents : ce que tu **envoies** (l'entrée : ta question, tes documents) et ce que le modèle **écrit** (la sortie). La sortie coûte environ 5 fois plus cher que l'entrée.

**Prix indicatifs au 10 septembre 2026, en euros par million de tokens** (conversion depuis le dollar à ≈ 0,92 €/$) :

| Taille du modèle | Exemple | Entrée / M tokens | Sortie / M tokens |
|---|---|---|---|
| Petit | Claude Haiku 4.5 | ≈ 0,90 € | ≈ 4,60 € |
| Moyen | Claude Sonnet 5 | ≈ 1,85 € | ≈ 9,20 € |
| Grand | Claude Opus 5 | ≈ 4,60 € | ≈ 23,00 € |
| Modèle ouvert, chez toi | Qwen 0.5B, Mistral 7B (via Ollama) | 0 € | 0 € |

⚠️ **Ces prix bougent, beaucoup et vite** — presque toujours à la baisse, et les noms de modèles changent tous les quelques mois. Ne les apprends pas par cœur : apprends à aller les lire sur la page de tarifs du fournisseur ([Anthropic](https://www.anthropic.com/pricing) · [OpenAI](https://openai.com/api/pricing/) · [Google](https://ai.google.dev/pricing)) et à refaire le calcul. C'est la méthode qui reste vraie, pas le nombre.

Faisons le calcul pour un vrai cas : **classer 10 000 documents**.

````python
# Tarifs en euros par MILLION de tokens (indicatifs, 10/09/2026)
tarifs = {
    "petit":  {"entree": 0.90, "sortie": 4.60},
    "moyen":  {"entree": 1.85, "sortie": 9.20},
    "grand":  {"entree": 4.60, "sortie": 23.00},
}

def espace(nombre):
    """Affiche 10000 comme « 10 000 » : plus lisible qu'un mur de chiffres."""
    return f"{nombre:,}".replace(",", " ")

def cout(nb_docs, tokens_entree, tokens_sortie, modele):
    """Coût total en euros pour `nb_docs` appels au modèle."""
    t = tarifs[modele]
    return (nb_docs * tokens_entree / 1e6) * t["entree"] + (nb_docs * tokens_sortie / 1e6) * t["sortie"]

# Un document de ~600 mots en entrée (800 tokens), une réponse courte en sortie (60 tokens)
NB_DOCS, ENTREE, SORTIE = 10_000, 800, 60

for modele in ("petit", "moyen", "grand"):
    print(f"{modele:6} : {cout(NB_DOCS, ENTREE, SORTIE, modele):7.2f} €  pour {espace(NB_DOCS)} documents")

facteur = cout(NB_DOCS, ENTREE, SORTIE, "grand") / cout(NB_DOCS, ENTREE, SORTIE, "petit")
print(f"\nLe grand modèle coûte ×{facteur:.1f} le petit, pour exactement le même travail.")
````

Une dizaine d'euros contre une cinquantaine : ce n'est pas énorme, mais multiplie par un traitement quotidien et tu passes de 300 € à 1 800 € par an. Le choix du modèle **est** une décision d'architecture, au même titre que le choix du langage.

La bonne question n'est jamais « quel est le meilleur modèle ? » mais « **le petit modèle suffit-il pour cette tâche ?** ». Pour classer, extraire, reformater : très souvent oui. Pour raisonner, écrire du code, analyser finement : souvent non.

### Le coût d'une heure de GPU

Si tu entraînes toi-même au lieu d'appeler une API, tu ne paies plus au token mais **à l'heure de machine**.

| Option | Prix | Ce que tu obtiens |
|---|---|---|
| Colab gratuit | **0 €** | Un GPU T4, quelques heures par jour, coupé sans préavis, aucune garantie |
| Colab Pro | ≈ 11 €/mois | Des GPU plus rapides, des sessions plus longues, un système d'unités de calcul ([détails](https://colab.research.google.com/signup)) |
| GPU loué à l'heure | ≈ 0,30 €/h (T4) à ≈ 2 €/h (A100) | Une machine à toi, facturée à la minute, tant qu'elle tourne ([exemple de grille](https://www.runpod.io/pricing)) |

Trois jours d'entraînement sur un A100 loué ≈ **150 €**. Et le GPU est facturé même pendant que tu réfléchis, que tu déjeunes ou que tu dors : c'est le piège numéro un.

### Pourquoi on prototype toujours sur un échantillon

D'où la règle d'or, valable pour le calcul comme pour les modèles :

> On fait tourner le pipeline complet sur **100 lignes** avant de le lancer sur 10 millions.

Trois raisons, dans cet ordre d'importance :

1. **L'erreur.** Ton code va planter à la ligne 47 000 sur une valeur manquante. Autant le découvrir en 2 secondes et pour 0,01 € qu'en 4 heures et pour 60 €.
2. **La boucle de retour.** Tu vas modifier ton prompt ou ton nettoyage une trentaine de fois. Trente essais à 3 secondes, c'est une matinée productive ; trente essais à 40 minutes, c'est une semaine perdue.
3. **L'estimation.** Un échantillon te donne le coût réel *avant* de t'engager. Une simple règle de trois, et tu sais si le projet coûte 12 € ou 12 000 €.

````python
ECHANTILLON = 100

cout_test = cout(ECHANTILLON, ENTREE, SORTIE, "moyen")
cout_total_estime = cout_test * NB_DOCS / ECHANTILLON

print(f"essai sur {ECHANTILLON} documents  : {cout_test:.3f} €")
print(f"estimation sur {espace(NB_DOCS)} docs : {cout_total_estime:.2f} €")
print(f"vérification (calcul direct)  : {cout(NB_DOCS, ENTREE, SORTIE, 'moyen'):.2f} €")
print()
print("Pour 20 centimes, tu sais combien coûtera le projet entier. C'est le meilleur")
print("rapport information / prix de tout le métier.")
````
