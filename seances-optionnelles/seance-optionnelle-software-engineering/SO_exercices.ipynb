## Préparation

Une seule cellule à lancer. Elle fabrique les données de tous les exercices (une liste de 2 millions de valeurs, un tableau NumPy, un DataFrame de 150 000 lignes, un fichier CSV), la grille de tarifs des modèles, et le helper `verifier`.

Elle prend quelques secondes.

````python
import time, sys, gc, os
import numpy as np
import pandas as pd

# ---------- Pour les exercices sur la vitesse ----------
N = 2_000_000
valeurs = [float(i) for i in range(N)]     # une liste Python : 2 millions de petits objets
tableau = np.arange(N, dtype=float)        # le même contenu, collé en mémoire

# ---------- Pour les exercices sur la mémoire ----------
generateur = np.random.default_rng(1)      # graine fixe → tout le monde a les mêmes données
n = 150_000
villes = ["Paris", "Lyon", "Marseille", "Lille", "Nantes", "Toulouse", "Nice", "Rennes"]
categories = ["sport", "musique", "cinéma", "jeux", "lecture"]

df = pd.DataFrame({
    "ville": generateur.choice(villes, n),
    "categorie": generateur.choice(categories, n),
    "note": generateur.integers(0, 11, n),          # entiers de 0 à 10
    "prix": (generateur.random(n) * 300).round(2),  # de 0 à 300 €
})

# ---------- Pour l'exercice sur la lecture par morceaux ----------
df.to_csv("donnees_exercices.csv", index=False)

def espace(nombre):
    """Affiche 150000 comme « 150 000 »."""
    return f"{nombre:,}".replace(",", " ")

# ---------- Vérification automatique des exercices ----------
def verifier(nom, condition):
    """Affiche ✅ ou ❌ sans jamais planter. `condition` = un booléen, ou une fonction sans argument."""
    try:
        ok = bool(condition() if callable(condition) else condition)
    except Exception as e:
        ok = False
        print(f"   (erreur pendant la vérification : {type(e).__name__} : {e})")
    print(("✅ " if ok else "❌ ") + nom + ("" if ok else "  → pas encore, relis l'énoncé et réessaie"))

print("Prêt :", espace(N), "valeurs,", espace(len(df)), "lignes,",
      f"{os.path.getsize('donnees_exercices.csv') / 1e6:.1f} Mo de CSV")
````

## Exercice 5 ⭐⭐ · Vectoriser un calcul de TVA

Les produits de `df` n'ont pas tous le même taux : **5,5 %** au-dessus de 200 €, **20 %** en dessous ou égal.

La version lente, avec une boucle, est déjà écrite. Écris la version **vectorisée** dans `ttc_vect`, sans aucune boucle, et calcule le facteur d'accélération dans `facteur_tva` (arrondi à l'entier).

Résultat attendu : les deux tableaux sont identiques, et `facteur_tva` vaut au moins 5.

<details><summary>Indice</summary>

`np.where(condition, valeur_si_vrai, valeur_si_faux)` applique un `if / else` à tout un tableau d'un coup. Ici : `np.where(prix > 200, prix * 1.055, prix * 1.20)`.

</details>

````python
# À toi
prix = df["prix"].to_numpy()

# --- version lente, déjà écrite ---
depart = time.perf_counter()
ttc_boucle = np.array([p * 1.055 if p > 200 else p * 1.20 for p in prix])
duree_lente = time.perf_counter() - depart

# --- version vectorisée, à toi ---
depart = time.perf_counter()
ttc_vect = None          # ← une seule ligne, sans boucle
duree_rapide = time.perf_counter() - depart

facteur_tva = None       # ← duree_lente / duree_rapide, arrondi

print(f"boucle    : {duree_lente:.4f} s")
print(f"vectorisé : {duree_rapide:.4f} s   → ×{facteur_tva}")
````

````python
verifier("Exercice 5 · ttc_vect", lambda: np.allclose(ttc_vect, ttc_boucle))
verifier("Exercice 5 · facteur_tva", lambda: facteur_tva >= 5)
````

<details><summary>Solution</summary>

```python

print(f"identiques : {np.allclose(ttc_vect, ttc_boucle)}")
print(f"×{facteur_tva} plus rapide")
```

La règle à retenir : dès que tu écris une boucle (ou une compréhension de liste) sur une colonne, il existe presque toujours une version vectorisée bien plus rapide.

## Exercice 10 ⭐⭐⭐ · Un allégeur automatique

Écris `alleger(donnees)` qui renvoie une **copie allégée** d'un DataFrame, sans jamais modifier l'original, en appliquant trois règles :

1. toute colonne de type `object` dont le nombre de valeurs différentes est inférieur à **la moitié** du nombre de lignes → `astype("category")` ;
2. toute colonne `int64` → `pd.to_numeric(..., downcast="integer")` (pandas choisit le plus petit type qui suffit) ;
3. toute colonne `float64` → `astype("float32")`.

Résultat attendu sur `df` : on passe d'environ 22 Mo à environ 1 Mo, soit **moins de 10 %** du poids d'origine, avec des valeurs inchangées et `note` devenue `int8`.

<details><summary>Indice</summary>

Parcours les colonnes avec `for colonne in donnees.columns:` et teste `str(donnees[colonne].dtype)`.
Pour compter les valeurs différentes : `donnees[colonne].nunique()`.
N'oublie pas `donnees = donnees.copy()` en première ligne.

</details>

````python
# À toi
def alleger(donnees):
    donnees = donnees.copy()
    # ← tes trois règles ici
    return donnees

df_alleg = alleger(df)

print(f"{df.memory_usage(deep=True).sum() / 1e6:.1f} Mo → {df_alleg.memory_usage(deep=True).sum() / 1e6:.2f} Mo")
print(df_alleg.dtypes)
````

````python
verifier("Exercice 10 · poids divisé par 10 au moins",
         lambda: df_alleg.memory_usage(deep=True).sum() < 0.10 * df.memory_usage(deep=True).sum())
verifier("Exercice 10 · note en int8", lambda: str(df_alleg["note"].dtype) == "int8")
verifier("Exercice 10 · prix en float32", lambda: str(df_alleg["prix"].dtype) == "float32")
verifier("Exercice 10 · valeurs inchangées",
         lambda: list(df_alleg["ville"]) == list(df["ville"]) and (df_alleg["note"] == df["note"]).all())
verifier("Exercice 10 · df d'origine intact", str(df["note"].dtype) == "int64")
````

<details><summary>Solution</summary>

```python
def alleger(donnees):
    donnees = donnees.copy()
    for colonne in donnees.columns:
        type_colonne = str(donnees[colonne].dtype)
        if type_colonne == "object" and donnees[colonne].nunique() < len(donnees) / 2:
            donnees[colonne] = donnees[colonne].astype("category")
        elif type_colonne == "int64":
            donnees[colonne] = pd.to_numeric(donnees[colonne], downcast="integer")
        elif type_colonne == "float64":
            donnees[colonne] = donnees[colonne].astype("float32")
    return donnees
```

Le test « valeurs inchangées » est le plus important : alléger, ce n'est pas perdre de l'information. `note` ne contient que des entiers de 0 à 10 — les 8 octets d'un `int64` étaient du gaspillage pur, `int8` suffit.

Attention tout de même à `float32` : on garde ~7 chiffres significatifs au lieu de ~15. C'est parfait pour des prix ou des mesures, à éviter pour des calculs scientifiques précis.

</details>

<details><summary>Solution</summary>

```python
def dependances(code, versions):
    modules = set()
    for ligne in code.splitlines():
        ligne = ligne.strip()
        if ligne.startswith("import "):
            for morceau in ligne[len("import "):].split(","):
                modules.add(morceau.strip().split()[0].split(".")[0])
        elif ligne.startswith("from "):
            modules.add(ligne.split()[1].split(".")[0])
    paquets = {NOMS_PYPI.get(m, m) for m in modules}
    return sorted(f"{p}=={versions[p]}" for p in paquets if p in versions)
```

Le filtre `if p in versions` fait deux choses d'un coup : il écarte les modules de la bibliothèque standard (`os`, `json`, `time`…), qui s'installent avec Python, et il t'oblige à connaître la version de ce que tu déclares. En vrai projet, `pip freeze > requirements.txt` fait ça pour toi — mais il liste *tout* l'environnement, y compris ce que tu n'utilises pas. Savoir ce qui doit vraiment y figurer reste ton travail.

</details>
