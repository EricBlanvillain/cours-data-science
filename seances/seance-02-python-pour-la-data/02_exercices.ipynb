## Exercice 3 ⭐⭐ · Un dictionnaire qui compte

La classe a voté pour le repas de fin d'année :
`votes = ["pizza", "sushi", "pizza", "tacos", "sushi", "pizza", "burger"]`.

- `comptes` : un dictionnaire plat → nombre de voix, rempli avec une boucle,
- `gagnant` : le plat qui a le plus de voix.

Résultat attendu : `{"pizza": 3, "sushi": 2, "tacos": 1, "burger": 1}` et `"pizza"`.

<details><summary>Indice</summary>

Dans la boucle : si le plat est déjà une clé (`if plat in comptes:`), on ajoute 1 ; sinon on crée la clé avec 1. Pour le gagnant, garde le meilleur trouvé jusqu'ici, comme le Pokémon le plus rapide de la leçon.

</details>

````python
# À toi
votes = ["pizza", "sushi", "pizza", "tacos", "sushi", "pizza", "burger"]

comptes = {}
gagnant = None

print(comptes, "->", gagnant)
````

````python
verifier("Exercice 3", comptes == {"pizza": 3, "sushi": 2, "tacos": 1, "burger": 1} and gagnant == "pizza")
````

<details><summary>Solution</summary>

```python
votes = ["pizza", "sushi", "pizza", "tacos", "sushi", "pizza", "burger"]

comptes = {}
for plat in votes:
    if plat in comptes:
        comptes[plat] += 1
    else:
        comptes[plat] = 1

gagnant = None
for plat in comptes:
    if gagnant is None or comptes[plat] > comptes[gagnant]:
        gagnant = plat

print(comptes, "->", gagnant)
```

C'est exactement ce que fait `value_counts()` en pandas, en une ligne : la boucle montre ce qu'il y a dessous.

</details>
