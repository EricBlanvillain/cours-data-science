## Exercice 6 ⭐⭐ · Une anomalie : les films hors norme

Avec la règle de la leçon (au-delà de 1,5 fois l'écart entre le 1er et le 3e quart), trouve les films **anormalement longs** :

````python
verifier("Exercice 6", (lambda q1, q3: limite_haute == q3 + 1.5 * (q3 - q1) and nb_trop_longs == (films["duree_min"] > limite_haute).sum() and plus_court == films.loc[films["duree_min"].idxmin(), "title"])(*films["duree_min"].quantile([0.25, 0.75])))
````

<details><summary>Solution</summary>

