pendant `tours` tours, on passe sur chaque fruit ; si la prédiction est fausse, on déplace **tout de suite** le seuil de `pas`
(vers le haut si le fruit était une pomme, vers le bas si c'était un melon). La fonction renvoie le **seuil final**.

Attendu : `seuil_final` entre 260 et 800 (les pommes en dessous, les melons au-dessus), donc `compter_erreurs(fruits, seuil_final) == 0`.

<details><summary>Indice</summary>

Deux boucles imbriquées : `for tour in range(tours):` puis `for poids, etiquette in fruits:`. À l'intérieur : `if predire(poids, seuil) != etiquette: seuil = seuil + pas if etiquette == "pomme" else seuil - pas`.

</details>

````python
# À toi
def apprendre(fruits, seuil=0, pas=50, tours=15):
    return seuil

seuil_final = apprendre(fruits)
print("seuil final :", seuil_final)
````

````python
verifier("Exercice 6", seuil_final is not None and 260 < seuil_final <= 800 and compter_erreurs(fruits, seuil_final) == 0)
````

<details><summary>Solution</summary>

```python
def apprendre(fruits, seuil=0, pas=50, tours=15):
    for tour in range(tours):
        for poids, etiquette in fruits:
            if predire(poids, seuil) != etiquette:
                seuil = seuil + pas if etiquette == "pomme" else seuil - pas
    return seuil

seuil_final = apprendre(fruits)
print("seuil final :", seuil_final)
```

## Exercice 7 ⭐⭐ · Le pas d'apprentissage

Que change le pas ? Écris `combien_de_tours(fruits, seuil, pas)` : on répète des tours (comme dans `apprendre`) **jusqu'à** obtenir 0 erreur,
et on renvoie le nombre de tours qu'il a fallu (avec une sécurité : jamais plus de 1000 tours).

Attendu : `tours_50 = 2` (pas de 50 g) et `tours_5` bien plus grand (pas de 5 g). Un petit pas est plus lent... mais un pas trop grand peut sauter par-dessus la bonne zone.

<details><summary>Indice</summary>

`tours = 0` puis `while compter_erreurs(fruits, seuil) > 0 and tours < 1000:` → un tour complet sur les fruits, puis `tours += 1`.

</details>

````python
# À toi
def combien_de_tours(fruits, seuil=0, pas=50):
    tours = 0
    return tours

tours_50 = combien_de_tours(fruits, seuil=0, pas=50)
tours_5 = combien_de_tours(fruits, seuil=0, pas=5)
print("pas de 50 g :", tours_50, "tours | pas de 5 g :", tours_5, "tours")
````

````python
verifier("Exercice 7", tours_50 == 2 and tours_5 > tours_50)
````

<details><summary>Solution</summary>

```python
def combien_de_tours(fruits, seuil=0, pas=50):
    tours = 0
    while compter_erreurs(fruits, seuil) > 0 and tours < 1000:
        for poids, etiquette in fruits:
            if predire(poids, seuil) != etiquette:
                seuil = seuil + pas if etiquette == "pomme" else seuil - pas
        tours += 1
    return tours

tours_50 = combien_de_tours(fruits, seuil=0, pas=50)
tours_5 = combien_de_tours(fruits, seuil=0, pas=5)
print("pas de 50 g :", tours_50, "tours | pas de 5 g :", tours_5, "tours")
```

```python
def evaluer_prompt(prompt):
    texte = prompt.lower()
    score = 0
    for regle, mots in INDICES.items():
        if any(m in texte for m in mots):
            score += 1
    return score
```

```python
def regles_manquantes(prompt):
    texte = prompt.lower()
    manquantes = []
    for regle, mots in INDICES.items():
```
