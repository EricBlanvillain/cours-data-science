## Exercice 6 ⭐⭐ · La machine de tri apprend son seuil

Personne ne règle la machine : elle apprend son seuil avec la recette **exemples → erreurs → corrections**. Écris
`apprendre(envois, seuil=0, pas=40, tours=15)` : pendant `tours` tours, on passe sur chaque envoi ; si le tri est faux, on déplace
**tout de suite** le seuil de `pas` (vers le haut si l'envoi était une lettre, vers le bas si c'était un colis). La fonction renvoie le **seuil final**.

Attendu : un `seuil_final` d'au moins 120 g (la lettre la plus lourde) et de moins de 300 g (le colis le plus léger), donc `compter_erreurs(envois, seuil_final) == 0`.

<details><summary>Indice</summary>

Deux boucles imbriquées : `for tour in range(tours):` puis `for poids, etiquette in envois:`. À l'intérieur : `if trier(poids, seuil) != etiquette: seuil = seuil + pas if etiquette == "lettre" else seuil - pas`.

</details>

````python
# À toi
def apprendre(envois, seuil=0, pas=40, tours=15):
    return seuil

seuil_final = apprendre(envois)
print("seuil final :", seuil_final)
````

````python
verifier("Exercice 6", seuil_final is not None and 120 <= seuil_final < 300 and compter_erreurs(envois, seuil_final) == 0)
````

<details><summary>Solution</summary>

```python
def apprendre(envois, seuil=0, pas=40, tours=15):
    for tour in range(tours):
        for poids, etiquette in envois:
            if trier(poids, seuil) != etiquette:
                seuil = seuil + pas if etiquette == "lettre" else seuil - pas
    return seuil

seuil_final = apprendre(envois)
print("seuil final :", seuil_final)
```

La machine s'arrête à 120 g : dès qu'elle ne se trompe plus, elle ne bouge plus.

## Exercice 7 ⭐⭐ · Trop petit pas, trop grand pas

Écris `combien_de_tours(envois, seuil, pas)` : on répète des tours (comme dans `apprendre`) **jusqu'à** obtenir 0 erreur,
et on renvoie le nombre de tours qu'il a fallu, avec une sécurité : jamais plus de 1000 tours.

Attendu : `tours_40 = 2` (pas de 40 g), `tours_4` bien plus grand (pas de 4 g), et `tours_400 = 1000` : avec un pas de 400 g, le seuil
saute par-dessus la bonne zone, de 0 à 400 et retour, sans jamais s'y poser.

<details><summary>Indice</summary>

`tours = 0` puis `while compter_erreurs(envois, seuil) > 0 and tours < 1000:` → un tour complet sur les envois, puis `tours += 1`.

</details>

````python
# À toi
def combien_de_tours(envois, seuil=0, pas=40):
    tours = 0
    return tours

tours_40 = combien_de_tours(envois, seuil=0, pas=40)
tours_4 = combien_de_tours(envois, seuil=0, pas=4)
tours_400 = combien_de_tours(envois, seuil=0, pas=400)
print("pas de 40 g :", tours_40, "| pas de 4 g :", tours_4, "| pas de 400 g :", tours_400)
````

````python
verifier("Exercice 7", tours_40 == 2 and tours_4 > tours_40 and tours_400 == 1000)
````

<details><summary>Solution</summary>

```python
def combien_de_tours(envois, seuil=0, pas=40):
    tours = 0
    while compter_erreurs(envois, seuil) > 0 and tours < 1000:
        for poids, etiquette in envois:
            if trier(poids, seuil) != etiquette:
                seuil = seuil + pas if etiquette == "lettre" else seuil - pas
        tours += 1
    return tours

tours_40 = combien_de_tours(envois, seuil=0, pas=40)
tours_4 = combien_de_tours(envois, seuil=0, pas=4)
tours_400 = combien_de_tours(envois, seuil=0, pas=400)
print("pas de 40 g :", tours_40, "| pas de 4 g :", tours_4, "| pas de 400 g :", tours_400)
```

Avec 4 g, il faut 17 tours. Un pas trop petit est lent, un pas trop grand ne s'arrête jamais : régler le pas fait partie du métier.

</details>

def score_prompt(prompt):

Un score ne dit pas quoi corriger. Écris `regles_manquantes(prompt)` qui renvoie la **liste des règles absentes**, dans l'ordre de `INDICES`,

Attendu : `conseils["Tom"] == ["contexte", "rôle", "exemple"]` · `conseils["Sacha"] == ["rôle", "exemple"]` · `conseils["Léa"] == []`.

<details><summary>Indice</summary>

</details>

````python
# À toi
def regles_manquantes(prompt):
    manquantes = []
    return manquantes

conseils = {nom: regles_manquantes(p) for nom, p in concours.items()}
for nom, manque in conseils.items():
    print(f"{nom:6s} : il manque {manque}")
````

def regles_manquantes(prompt):
