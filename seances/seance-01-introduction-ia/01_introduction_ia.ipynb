## 2. Les trois cercles : IA, machine learning, deep learning

Imagine trois cercles emboîtés, comme des poupées russes :

- **IA** (le grand cercle, années 1950) : tout programme qui imite une capacité humaine. Un GPS, un programme d'échecs, un chatbot.
- **Machine learning** (au milieu, années 1990-2000) : le programme **apprend une règle à partir d'exemples** au lieu qu'on la lui écrive. Filtre anti-spam, recommandations.
- **Deep learning** (au centre, depuis 2012) : du machine learning avec des **réseaux de neurones** géants, inspirés (de loin) du cerveau. Reconnaissance d'images, voix, traduction.

L'**IA générative** (ChatGPT, Midjourney, Suno) est née dans le petit cercle : ce sont des réseaux de neurones énormes entraînés à **produire** du texte, des images ou du son, au lieu de seulement reconnaître.

## 3. Comment une machine « apprend »

Recette universelle du machine learning : **des exemples, des erreurs, des corrections**.

Prenons 20 fruits pesés sur une balance : des pommes et des melons. On veut une règle du genre « si le poids dépasse un **seuil**, c'est un melon ».
Au lieu de choisir le seuil nous-mêmes, la machine part d'un seuil au hasard, regarde ses erreurs, et **déplace le seuil** un peu à chaque erreur.
C'est exactement l'idée derrière les réseaux de neurones, en version miniature.

````python
random.seed(3)
# 10 pommes (120-260 g) et 10 melons (700-2500 g) : (poids, étiquette)
fruits = [(random.randint(120, 260), "pomme") for _ in range(10)]
fruits += [(random.randint(700, 2500), "melon") for _ in range(10)]
random.shuffle(fruits)

for poids, etiquette in fruits[:5]:
    print(poids, "g ->", etiquette)
print("...")
````

````python
def predire(poids, seuil):
    return "melon" if poids > seuil else "pomme"

def compter_erreurs(fruits, seuil):
    return sum(1 for poids, etiquette in fruits if predire(poids, seuil) != etiquette)

def apprendre(fruits, seuil=0, pas=50, tours=15):
    """À chaque erreur, on pousse le seuil dans la bonne direction. Renvoie l'historique."""
    historique = []
    for tour in range(tours):
        for poids, etiquette in fruits:
            if predire(poids, seuil) != etiquette:
                seuil = seuil + pas if etiquette == "pomme" else seuil - pas
        historique.append((seuil, compter_erreurs(fruits, seuil)))
        print(f"tour {tour + 1:2d} : seuil = {seuil:5d} g, erreurs = {historique[-1][1]}")
    return historique

historique = apprendre(fruits)
````

````python
seuils = [s for s, e in historique]
erreurs = [e for s, e in historique]

fig, (g1, g2) = plt.subplots(1, 2, figsize=(11, 4))
g1.plot(range(1, len(erreurs) + 1), erreurs, marker="o")
g1.set_xlabel("tour d'apprentissage"); g1.set_ylabel("nombre d'erreurs"); g1.set_title("Les erreurs diminuent")

for poids, etiquette in fruits:
    g2.scatter(poids, 0, color="red" if etiquette == "pomme" else "green", s=60)
g2.axvline(seuils[-1], color="black", linestyle="--", label=f"seuil appris = {seuils[-1]} g")
g2.set_yticks([]); g2.set_xlabel("poids (g)"); g2.set_title("Rouge = pomme, vert = melon"); g2.legend()
plt.show()
````

## 6. Projet : le concours de prompts

Un **prompt**, c'est ce que tu écris à l'IA. Un mauvais prompt = une réponse floue. Les 5 règles d'un bon prompt :

1. **Contexte** : qui tu es, pour quoi faire (« je suis en 3e, je prépare un exposé »).
2. **Rôle** : « tu es un prof de SVT patient ».
3. **Tâche précise** : un verbe, un objet (« explique la photosynthèse »).
4. **Format** : « en 5 points », « un tableau », « 3 phrases », « du code Python ».
5. **Exemple** : montre ce que tu attends (« par exemple : ... »).

La cellule suivante définit une **grille d'évaluation automatique** : elle cherche des indices de chaque règle dans ton prompt et donne un score sur 5.
Elle n'est pas parfaite (c'est un programme à règles, pas une IA !), mais elle te dit ce qui manque.

````python
INDICES = {
    "contexte": ["je suis", "je prépare", "pour mon", "pour ma", "j'ai", "nous sommes", "dans le cadre", "mon prof", "ma classe"],
    "rôle": ["tu es", "agis comme", "en tant que", "joue le rôle", "imagine que tu es"],
    "tâche": ["explique", "écris", "résume", "propose", "crée", "liste", "génère", "traduis", "corrige", "compare", "rédige", "donne", "invente"],
    "format": ["liste", "tableau", "phrases", "mots maximum", "paragraphe", "json", "étapes", "points", "lignes", "titre", "code python"],
    "exemple": ["exemple", "comme ceci", "par ex", "modèle :", "voici", "du genre"],
}

def evaluer_prompt(prompt):
    texte = prompt.lower()
    score = 0
    for regle, mots in INDICES.items():
        trouve = [m for m in mots if m in texte]
        present = len(trouve) > 0
        score += present
        print("✅" if present else "❌", f"{regle:9s}", f"(indice : {trouve[0]})" if present else "-> il manque quelque chose ici")
    print(f"\nScore : {score} / 5")
    return score

evaluer_prompt("explique la photosynthèse")
````
