## Préparation

Une seule cellule à lancer : elle crée `verifier`, qui affiche ✅ ou ❌ pour chaque exercice.

````python
resultats = {}

def verifier(nom, condition):
    """Affiche ✅ ou ❌ sans jamais lever d'exception (condition = un booléen, ou une fonction sans argument)."""
    try:
        ok = bool(condition() if callable(condition) else condition)
    except Exception:
        ok = False
    resultats[nom] = ok
    print(("✅ " if ok else "❌ ") + nom + ("" if ok else " — pas encore, relis l'énoncé ou ouvre l'indice."))

print("Prêt !")
````

## Exercice 2 ⭐ · Une variable qui change

Ta cagnotte contient **50 €**. Pour ton anniversaire, on t'offre **20 €** ; ensuite tu achètes un jeu à **35 €**.

Mets la cagnotte à jour **dans la même variable**, une ligne par événement : `cagnotte = cagnotte + ...`.

Résultat attendu : `cagnotte` vaut `35`.

<details><summary>Indice</summary>

À droite du `=`, Python calcule avec l'ancienne valeur ; à gauche, il range le résultat dans la même boîte.

</details>

````python
# À toi
cagnotte = 50
# deux lignes : l'anniversaire, puis l'achat du jeu

print("Il reste", cagnotte, "€")
````

````python
verifier("Exercice 2", cagnotte == 35)
````

<details><summary>Solution</summary>

```python
cagnotte = 50
cagnotte = cagnotte + 20
cagnotte = cagnotte - 35

print("Il reste", cagnotte, "€")
```

Une variable n'est pas figée : c'est une boîte dont on peut changer le contenu. `cagnotte += 20` est le raccourci de `cagnotte = cagnotte + 20`.

</details>

## Exercice 3 ⭐ · Partager équitablement

**100 bonbons** à partager entre **7 amis**, sans couper de bonbon.

- `chacun` : combien de bonbons reçoit chaque ami (division **entière**, `//`),
- `reste` : combien il en reste (le **reste** de la division, `%`).

Résultat attendu : `14` et `2`.

<details><summary>Indice</summary>

`17 // 5` vaut `3` (on garde la partie entière) et `17 % 5` vaut `2` (ce qui reste).

</details>

````python
# À toi
bonbons = 100
amis = 7

chacun = None
reste = None

print(chacun, "bonbons chacun, il en reste", reste)
````

````python
verifier("Exercice 3", chacun == 14 and reste == 2)
````

<details><summary>Solution</summary>

```python
bonbons = 100
amis = 7

chacun = bonbons // amis
reste = bonbons % amis

print(chacun, "bonbons chacun, il en reste", reste)
```

Vérification de tête : 7 × 14 = 98, et 98 + 2 = 100.

</details>

## Exercice 5 ⭐⭐ · Du texte

Le pseudo d'une joueuse est `"luna_star"`. Calcule :

- `en_majuscules` : le pseudo en MAJUSCULES (`.upper()`),
- `longueur` : son nombre de caractères (`len`),

Le `+` entre deux textes les colle bout à bout. Attention à l'espace : `pseudo + " a rejoint la partie"`.

## Exercice 7 ⭐⭐ · Une moyenne arrondie

Tes trois notes du trimestre : **14**, **11.5** et **16**. Calcule `moyenne`, arrondie à **1 décimale** avec `round(nombre, 1)`.

Résultat attendu : `13.8`.

<details><summary>Indice</summary>

Les parenthèses d'abord : `(14 + 11.5 + 16) / 3`, puis `round(..., 1)`. Sans parenthèses, Python diviserait seulement la dernière note.

</details>

````python
# À toi
note1 = 14
note2 = 11.5
note3 = 16

moyenne = None

print("Moyenne :", moyenne)
````

````python
verifier("Exercice 7", moyenne == 13.8)
````

<details><summary>Solution</summary>

```python
note1 = 14
note2 = 11.5
note3 = 16

moyenne = round((note1 + note2 + note3) / 3, 1)

print("Moyenne :", moyenne)
```

En Python, la virgule des décimales s'écrit avec un point : `11.5`, jamais `11,5`.

</details>

## Exercice 9 ⭐⭐ · Combien de photos ?

Ta carte mémoire fait **64 Go** et une photo pèse **4 Mo**. On compte 1 Go = 1 000 Mo.

- `capacite_mo` : la capacité de la carte en Mo,
- `nb_photos` : le nombre de photos **complètes** qu'elle peut contenir.

Puis une question bonus sur les puissances : `kilo_info` vaut `2 ** 10` (2 multiplié 10 fois par lui-même), le « kilo » des informaticiens.

Une photo ne se coupe pas : division entière `//`. Et `**` est l'opérateur « puissance ».

````python
# À toi
carte_go = 64
photo_mo = 4

capacite_mo = None
nb_photos = None
kilo_info = None

print(capacite_mo, nb_photos, kilo_info)
````

## Exercice 10 ⭐⭐⭐ · Une barre de progression en texte

Un téléchargement en est à **70 %**. On veut l'afficher comme une barre de 10 cases :

`■■■■■■■□□□ 70 %`

- `pleines` : le nombre de cases pleines (une case pour chaque tranche de 10 %),
- `barre` : le texte des 10 cases, pleines `■` puis vides `□`,
- `affichage` : la barre, une espace, puis le pourcentage suivi de ` %`.

Nouveau : un texte multiplié par un nombre se **répète**. `"ab" * 3` vaut `"ababab"`.

<details><summary>Indice</summary>

`pleines = progression // 10`, puis `"■" * pleines + "□" * (10 - pleines)`. Pour l'affichage, une f-string.

</details>

````python
# À toi
progression = 70

pleines = None
barre = None
affichage = None

print(affichage)
````

````python
verifier("Exercice 10", pleines == 7 and barre == "■■■■■■■□□□" and affichage == "■■■■■■■□□□ 70 %")
````

<details><summary>Solution</summary>

```python
progression = 70

pleines = progression // 10
barre = "■" * pleines + "□" * (10 - pleines)
affichage = f"{barre} {progression} %"

print(affichage)
```

Change `progression` en `30` et relance : la barre se redessine toute seule. C'est l'intérêt de tout calculer à partir d'une variable.

</details>

## Exercice 11 ⭐⭐⭐ · L'heure du match à Tokyo

Un match commence à **21 h** à Paris. Tokyo a **7 heures d'avance**. Quelle heure est-il à Tokyo ?

- `heure_tokyo` : l'heure à Tokyo, entre 0 et 23 (une journée fait 24 heures : le reste `%` remet le compteur à zéro après minuit),
- `lendemain` : `True` si à Tokyo on est déjà le jour suivant (21 + 7 dépasse-t-il 23 ?),
- `affichage` : l'heure sur deux chiffres, comme une horloge : `"04 h 00"`.

Nouveau : dans une f-string, `{nombre:02d}` écrit le nombre sur 2 chiffres, avec un zéro devant si besoin.

<details><summary>Indice</summary>

`(21 + 7) % 24` vaut `4`. Pour l'affichage : `f"{heure_tokyo:02d} h 00"`.

</details>

````python
# À toi
heure_paris = 21
decalage = 7

heure_tokyo = None
lendemain = None
affichage = None

print(affichage, "· le lendemain :", lendemain)
````

````python
verifier("Exercice 11", heure_tokyo == 4 and lendemain is True and affichage == "04 h 00")
````

<details><summary>Solution</summary>

```python
heure_paris = 21
decalage = 7

heure_tokyo = (heure_paris + decalage) % 24
lendemain = heure_paris + decalage >= 24
affichage = f"{heure_tokyo:02d} h 00"

print(affichage, "· le lendemain :", lendemain)
```

Le reste de la division par 24, c'est exactement ce que fait une horloge : après 23 h, on repart à 0.

</details>

## Exercice 12 ⭐⭐⭐ · Défi : le budget du voyage

Trois amis partent en voiture : **450 km**, une voiture qui consomme **6 litres aux 100 km**, l'essence à **1,85 € le litre**.

- `litres` : l'essence nécessaire,
- `cout` : le prix total de l'essence, arrondi au centime (`round(..., 2)`),
- `par_personne` : la part de chacun, arrondie au centime,
- `resume` : le texte exact `"27.0 L d'essence, 49.95 €, soit 16.65 € chacun."`, construit avec une f-string à partir des trois variables.

Tout se calcule à partir des variables du haut : aucun nombre tapé à la main dans les calculs.

<details><summary>Indice</summary>

`litres = distance / 100 * conso`. Le coût : `round(litres * prix_litre, 2)`. Puis la part : `round(cout / amis, 2)`.

</details>

````python
# À toi
distance = 450
conso = 6
prix_litre = 1.85
amis = 3

litres = None
cout = None
par_personne = None
resume = None

print(resume)
````

````python
verifier("Exercice 12", litres == 27.0 and cout == 49.95 and par_personne == 16.65 and resume == "27.0 L d'essence, 49.95 €, soit 16.65 € chacun.")
````

<details><summary>Solution</summary>

```python
distance = 450
conso = 6
prix_litre = 1.85
amis = 3

litres = distance / 100 * conso
cout = round(litres * prix_litre, 2)
par_personne = round(cout / amis, 2)
resume = f"{litres} L d'essence, {cout} €, soit {par_personne} € chacun."

print(resume)
```

Pourquoi arrondir ? Ici `16.65` tombe juste, mais ce n'est pas toujours le cas : l'ordinateur calcule en binaire, et `0.1 + 0.2` donne `0.30000000000000004`. Avec de l'argent, on arrondit toujours au centime.

</details>
