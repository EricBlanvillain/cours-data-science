Écris `coups_malins(mystere, bas=1, haut=50)` qui renvoie la **liste des propositions** d'un joueur qui coupe l'intervalle en deux à chaque coup :
il propose le milieu `(bas + haut) // 2` ; si c'est trop petit, `bas` devient `milieu + 1` ; si c'est trop grand, `haut` devient `milieu - 1` ; on s'arrête quand c'est trouvé.

Attendu : `coups_malins(37)` → `[25, 38, 31, 34, 36, 37]` et `coups_malins(1)` → `[25, 12, 6, 3, 1]`.
Bonus vérifié : entre 1 et 50, cette stratégie ne dépasse jamais 6 coups.

<details><summary>Indice</summary>

Une boucle `while bas <= haut:` avec `milieu = (bas + haut) // 2`, `coups.append(milieu)`, puis `if milieu == mystere: break`.

</details>

````python
# À toi
def coups_malins(mystere, bas=1, haut=50):
    coups = []
    return coups

print(coups_malins(37), coups_malins(1))
````

````python
verifier("Exercice 10", coups_malins(37) == [25, 38, 31, 34, 36, 37] and coups_malins(1) == [25, 12, 6, 3, 1] and all(len(coups_malins(m)) <= 6 for m in range(1, 51)))
````

<details><summary>Solution</summary>

```python
def coups_malins(mystere, bas=1, haut=50):
    coups = []
    while bas <= haut:
        milieu = (bas + haut) // 2
        coups.append(milieu)
        if milieu == mystere:
            break
        elif milieu < mystere:
            bas = milieu + 1
        else:
            haut = milieu - 1
    return coups

print(coups_malins(37), coups_malins(1))
```

        return "égalité"

## Exercice 12 ⭐⭐⭐ · Défi : le pendu sans clavier

Écris `pendu(mot, lettres)` qui renvoie **deux valeurs** : le mot masqué (chaque lettre est affichée si elle est dans `lettres`, sinon `_`)
et le nombre d'erreurs (les lettres proposées qui ne sont pas dans le mot).

Exemples : `pendu("python", ["p", "y", "a", "o"])` → `("py__o_", 1)` · `pendu("data", ["a", "z", "t", "d"])` → `("data", 1)` · `pendu("ia", [])` → `("__", 0)`.

Tu combines ici : une boucle sur les lettres d'un mot, `in`, un compteur, la construction d'un texte et `return a, b`.

<details><summary>Indice</summary>

Commence par `affichage = ""`, puis pour chaque `lettre in mot` : `affichage += lettre if lettre in lettres else "_"`. Les erreurs : une deuxième boucle sur `lettres` avec `if lettre not in mot`.

</details>

````python
# À toi
def pendu(mot, lettres):
    affichage = ""
    erreurs = 0
    return affichage, erreurs

print(pendu("python", ["p", "y", "a", "o"]))
print(pendu("data", ["a", "z", "t", "d"]))
print(pendu("ia", []))
````

````python
verifier("Exercice 12", pendu("python", ["p", "y", "a", "o"]) == ("py__o_", 1) and pendu("data", ["a", "z", "t", "d"]) == ("data", 1) and pendu("ia", []) == ("__", 0))
````

<details><summary>Solution</summary>

```python
def pendu(mot, lettres):
    affichage = ""
    for lettre in mot:
        if lettre in lettres:
            affichage += lettre
        else:
            affichage += "_"
    erreurs = 0
    for lettre in lettres:
        if lettre not in mot:
            erreurs += 1
    return affichage, erreurs

print(pendu("python", ["p", "y", "a", "o"]))
print(pendu("data", ["a", "z", "t", "d"]))
print(pendu("ia", []))
```

</details>
