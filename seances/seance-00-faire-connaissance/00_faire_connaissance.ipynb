## Préparation

Rien à installer aujourd'hui. La cellule prépare les six questions du petit bilan (section 2).

```python
# Le petit bilan de la section 2 : six bouts de code à lire (pas à écrire)
BILAN = [
    ("x = 7\nx = x + 3\nprint(x)", ["7", "10", "x + 3", "une erreur"], "b"),
    ('print("3" + "4")', ["34", "7", "3 + 4", "une erreur"], "a"),
    ("print(23 // 4, 23 % 4)", ["5 3", "5.75 0", "3 5", "une erreur"], "a"),
    ("age = 15\nif age >= 18:\n    print(\"majeur\")\nelse:\n    print(\"mineur\")", ["rien", "mineur", "majeur", "une erreur"], "b"),
    ('jeux = ["Zelda", "Mario", "Tetris"]\nprint(jeux[1])', ["Zelda", "Tetris", "Mario", "une erreur"], "c"),
    ("total = 0\nfor n in [2, 4, 6]:\n    total = total + n\nprint(total)", ["6", "246", "une erreur", "12"], "d"),
]

print("Prêt ! Python répond, on peut commencer.")
```

## 1. Python répond ?

Un notebook, c'est une suite de **cellules** : du texte (comme ici) ou du code. Quand tu exécutes une cellule de code,
Python lit les lignes de haut en bas et affiche le résultat juste en dessous.

Exécute la cellule suivante. Si tu vois « Bonjour », tout fonctionne.

```python
print("Bonjour ! Python fonctionne.")
print(2 + 2)
```

Une **variable** range une valeur sous un nom. Remplace le prénom et l'âge par les tiens, puis relance.

```python
# À toi : remplace par ton prénom et ton âge, puis relance
prenom = "Alex"
age = 15
print("Je m'appelle", prenom, "et j'ai", age, "ans.")
```

Python calcule : `+ - * /`, et deux opérations bien pratiques, la division entière `//` et le reste `%`.

```python
print(2026 - age)
print(17 // 5, 17 % 5)
```

Du texte : `len` compte les caractères, `.upper()` met en majuscules.

```python
print(len(prenom), prenom.upper())
```

Une **f-string** glisse des variables dans une phrase : un `f` devant les guillemets, les variables entre accolades.

```python
print(f"Dans 10 ans, {prenom} aura {age + 10} ans.")
```

Une comparaison répond `True` (vrai) ou `False` (faux).

```python
print(age >= 18, prenom == "Alex")
```

## 2. Petit bilan : ce que tu sais déjà

Six bouts de code. Pour chacun, devine ce qu'il affiche (`a`, `b`, `c` ou `d`), **sans l'exécuter**. Si tu n'as jamais vu ça, réponds au hasard ou mets `"?"` : c'est justement ce qu'on veut savoir. Rien n'est noté, et on ne corrige pas maintenant : tout ça s'apprend à la séance 2.

```python
for numero, (code, choix, _) in enumerate(BILAN, start=1):
    print(f"--- Question {numero} ---\n{code}")
    print("   ".join(f"{l}. {c}" for l, c in zip("abcd", choix)), "\n")
```

Tes six réponses, dans l'ordre.

```python
# À toi
mes_reponses = ["?", "?", "?", "?", "?", "?"]
```

Le bilan : combien de bonnes réponses. Le formateur s'en sert pour adapter les prochaines séances.

```python
bonnes = sum(r == bonne for r, (_, _, bonne) in zip(mes_reponses, BILAN))
print(f"{bonnes} bonne(s) réponse(s) sur {len(BILAN)}")
```
