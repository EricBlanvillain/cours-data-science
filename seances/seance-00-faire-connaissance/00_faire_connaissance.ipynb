## 6. Projet : le jeu du nombre mystère

L'ordinateur choisit un nombre entre 1 et 50. Le joueur propose des nombres, le programme répond « plus grand » ou « plus petit »
jusqu'à ce qu'il trouve. C'est un projet complet : une boucle, des conditions, un compteur.

On commence par une version **sans clavier** : les propositions du joueur sont écrites à l'avance dans une liste.
Ça permet de tester le programme en un clic. La version interactive (avec `input()`) est tout en bas du notebook.

````python
def jouer(mystere, propositions):
    """Joue une partie avec des coups prédéfinis. Renvoie le nombre d'essais."""
    essais = 0
    for proposition in propositions:
        essais += 1
        if proposition < mystere:
            print(f"{proposition} -> plus grand !")
        elif proposition > mystere:
            print(f"{proposition} -> plus petit !")
        else:
            print(f"{proposition} -> bravo, trouvé en {essais} essais.")
            return essais
    print("Pas trouvé... le nombre était", mystere)
    return essais

jouer(mystere=37, propositions=[25, 40, 30, 35, 37])
````

**Exercice** : l'ordinateur tire le nombre au hasard avec `random.randint(1, 50)`. Écris une liste de propositions
**intelligente** : à chaque coup, propose le milieu de l'intervalle possible (25, puis 12 ou 37, ...).
Combien d'essais au maximum pour trouver à coup sûr entre 1 et 50 ?

````python
random.seed(7)                 # avec seed, le hasard est le même à chaque exécution
mystere = random.randint(1, 50)

# À toi : remplace cette liste par une stratégie « milieu de l'intervalle »
mes_coups = [10, 20, 30, 40, 50, 1]
jouer(mystere, mes_coups)
````

<details><summary>Solution</summary>

```python
# Stratégie automatique : on coupe l'intervalle en deux à chaque coup (recherche dichotomique)
def coups_malins(mystere, bas=1, haut=50):
    coups = []
    while bas <= haut:
        milieu = (bas + haut) // 2
        coups.append(milieu)
        if milieu == mystere:
            break
        elif milieu < mystere:
            bas = milieu + 1
        else:
            haut = milieu - 1
    return coups

jouer(mystere, coups_malins(mystere))   # 6 essais maximum entre 1 et 50
```
