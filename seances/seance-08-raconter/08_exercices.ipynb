    une = boutique[boutique["plateforme"] == p]
    dash.add_trace(go.Scatter(x=une["mois_num"], y=une["ventes"], name=p, mode="lines+markers"))

dash.update_layout(updatemenus=[dict(buttons=boutons, x=0, y=1.2)], title="Les ventes de 2025 : choisis la plateforme")

if dash is not None:
    dash.show()

## Exercice 10 ⭐⭐⭐ · Le pitch en trois temps

Calcule d'abord deux chiffres pour le gérant :

- `part_decembre` : la part des ventes de l'année faites en décembre, en %, arrondie à 1 décimale ;
- `croissance_pc` : la hausse des ventes PC entre janvier et décembre, en %, arrondie à l'unité.

Puis écris le `pitch` : un dictionnaire avec `constat`, `preuve` et `recommandation`. Le constat doit citer l'un des deux chiffres
(écrit avec une virgule ou un point), et chaque case fait au moins 30 caractères.

Résultat attendu : `18.0` et `249`.

<details><summary>Indice</summary>

La part : `boutique[boutique["mois"] == "déc"]["ventes"].sum() / boutique["ventes"].sum() * 100`. La hausse : `(décembre / janvier - 1) * 100`, avec les ventes PC de ces deux mois.

</details>

````python
# À toi
part_decembre = None
croissance_pc = None
pitch = {"constat": "...", "preuve": "...", "recommandation": "..."}

print(part_decembre, croissance_pc)
for etape, texte in pitch.items():
    print(f"{etape.upper():15s} {texte}")
````

````python
verifier("Exercice 10", part_decembre == 18.0 and croissance_pc == 249 and all(len(t) >= 30 and not t.startswith("...") for t in pitch.values()) and any(str(x) in pitch["constat"] or str(x).replace(".", ",") in pitch["constat"] for x in (part_decembre, croissance_pc)))
````

<details><summary>Solution</summary>

```python
pc = boutique[boutique["plateforme"] == "PC"].set_index("mois")["ventes"]
part_decembre = round(boutique[boutique["mois"] == "déc"]["ventes"].sum() / boutique["ventes"].sum() * 100, 1)
croissance_pc = round((pc["déc"] / pc["janv"] - 1) * 100)
pitch = {
    "constat": f"Décembre fait à lui seul {str(part_decembre).replace('.', ',')} % des ventes de l'année, et le PC grimpe de {croissance_pc} % entre janvier et décembre.",
    "preuve": "La courbe des ventes mois par mois, axe à zéro : les trois plateformes bondissent en novembre et décembre, le PC monte toute l'année.",
    "recommandation": "Commander le stock de Noël dès septembre, et donner plus de place aux jeux PC en rayon dès le printemps.",
}

print(part_decembre, croissance_pc)
for etape, texte in pitch.items():
    print(f"{etape.upper():15s} {texte}")
```

Le chiffre dans le constat, c'est ce qui le rend vérifiable : le gérant peut demander « montre-moi », et la preuve suit.

</details>
